```python
# IMPORTAÇÃO DE BIBLIOTECAS

import pandas as pd
import numpy as np

# CARREGAMENTO DE DADOS

gini = pd.read_csv("../data/raw/economic-inequality-index.csv")
gdp = pd.read_csv("../data/raw/gdp-per-capita.csv")
idh = pd.read_csv("../data/raw/human-development-index.csv")
edu = pd.read_csv("../data/raw/years-of-schooling.csv")
```

```python
# TRATAMENTO DE DADOS - RENOMEAR COLUNAS

gini = gini.rename(columns={
    'Entity' : 'country',
    'Year' : 'year',
    'Gini coefficient' : 'gini',
    'World region according to OWID' : 'region'
})
```

```python
# TRATAMENTO DE DADOS - CONVERSÃO DE TIPOS

gini['year'] = pd.to_numeric(gini['year'], errors='coerce')
gini['gini'] = pd.to_numeric(gini['gini'], errors='coerce')
```

```python
# TRATAMENTO DE DADOS - SELECIONAR COLUNAS RELEVANTES

gini = gini[['country', 'year', 'gini', 'region']]
```

```python
# TRATAMENTO DE DADOS - CONFIRMAR SE OS OUTROS DATASETS ESTÃO NO FORMATO LONG

gdp.head()
```

```python
idh.head()
```

```python
edu.head()
```

```python
# TRATAMENTO DE DADOS - RENOMEAR COLUNAS DOS OUTROS DATASETS

gdp = gdp.rename(columns={
    'Entity' : 'country',
    'Year' : 'year',
    'GDP per capita' : 'gdp_per_capita'
})
```

```python
idh = idh.rename(columns={
    'Entity' : 'country',  
    'Year' : 'year',
    'Human Development Index' : 'idh'
})
```

```python
edu = edu.rename(columns={
    'Entity' : 'country',
    'Year' : 'year',
    'Both genders' : 'education'
})
```

```python
# TRATAMENTO DE DADOS - SELECIONAR TODAS AS COLUNAS

gini = gini[['country', 'year', 'gini', 'region']]
gdp = gdp[['country', 'year', 'gdp_per_capita']]
idh = idh[['country', 'year', 'idh']]
edu = edu[['country', 'year', 'education']]
```

```python
# GARANTIR TIPOS

for df in [gini, gdp, idh, edu]:
    df['year'] = pd.to_numeric(df['year'], errors='coerce')
```

```python
# TRATAMENTO DE DADOS - MERGE DOS DATASETS

df = gini.merge(gdp, on=['country', 'year'], how='inner')
df = df.merge(idh, on=['country', 'year'], how='inner')
df = df.merge(edu, on=['country', 'year'], how='inner')
```

```python
# TRATAMENTO DE DADOS - LIMPEZA FINAL

df = df.dropna()
```

```python
df = df.reset_index(drop=True)
```
